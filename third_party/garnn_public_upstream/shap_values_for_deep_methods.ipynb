```python
import shap
import warnings
warnings.simplefilter('ignore')

def get_index(name, lst):
    for i, l in enumerate(lst):
        if name == l:
            return i

    

def average_weights(pid2model):
    """
    Returns the average of the weights.
    """
    pid_list = list(pid2model.keys())

    pid2state_dict = {pid: pid2model[pid].state_dict() for pid in pid_list}

    w_avg = copy.deepcopy(pid2state_dict[pid_list[0]])

    for key in w_avg.keys():
        for pid in pid_list[1:]:
            w_avg[key] += pid2state_dict[pid][key]
        w_avg[key] = torch.div(w_avg[key], len(pid_list))
    return w_avg

method2allimport = {}
dataset = 'ohio'
method = 'gatv2_exp_layer_1'

dataset2example_idx = {
    'ohio': 150 + 258,
    'arises': 0 + 244, 
    'shanghai_t1dm': 0 + 113,
    'shanghai_t2dm' : 0 + 74,  
}
dataset2example_pid = {
    'ohio': 591, 
    'arises' : 1005,
    'shanghai_t1dm' : 1001,
    'shanghai_t2dm' : 2030,

}

# for dataset in ['ohio', 'arises']:
for dataset in ['ohio', 'arises', 'shanghai_t1dm', 'shanghai_t2dm']:
    seed = 1
    from gat_mem_org_log.train import *
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False
    os.environ['PYTHONHASHSEED'] = str(seed)

    version = 'nips_23'

    CONF = select_conf(dataset=dataset, version=version, seed=seed, method=method, hidden_dim=mthd_dt2hidden_dim[(method, dataset)])

    log = Log(CONF['dataset'] + '_' + CONF['model_name'] + '_' + CONF['comments'], CONF)
    data_proc = DataProcessor(CONF)



    pid_list = list(data_proc.train_pid2data_npy.keys())
    update = Update(data_proc, CONF)
    attris_name = ['glucose_level'] + CONF['attri_list'] + CONF['time_attri_list']

    pid2model = {}
    pid_shap_values = []
    example_shap_values = None
    pid_X_trains = []
    for pid in pid_list:

        pid2model[pid] = get_model(CONF)
        pid2model[pid] = log.load_models(pid2model[pid], CONF, pid)

        pid2model[pid].to(CONF['device'])


        G, attris, time_attris, y, target_time = data_proc.get_all_train(CONF['device'], CONF['seq_len'],  CONF['attri_list'], CONF['time_attri_list'], pid )

        X_train = torch.concatenate([G.unsqueeze(dim=-1), attris, time_attris], dim=-1)
        try:
            X_train_idx = np.random.choice(X_train.shape[0], 100, replace=False)
        except:
            continue

        explainer = shap.DeepExplainer(pid2model[pid], X_train[X_train_idx])
        shap_values = explainer.shap_values(X_train[X_train_idx])
        pid_shap_values.append(shap_values)
        pid_X_trains.append(X_train[X_train_idx])


        # save test examples
        if pid == dataset2example_pid[dataset]:
            G, attris, time_attris, y, target_time, seq_st_ed = data_proc.get_test(CONF['device'], CONF['seq_len'], CONF['attri_list'], CONF['time_attri_list'], pid)
            X_test = torch.cat([G.unsqueeze(dim=-1), attris, time_attris], dim=-1)
            j = dataset2example_idx[dataset]
            example_shap_values = explainer.shap_values(X_test[j][None, ...]) # 1 T C
            
        # end saving

    save_path = os.path.join(CONF['log_path'], 'SHAP')
    if not os.path.exists(save_path):
        os.makedirs(save_path)
    shap_values = np.concatenate(pid_shap_values, axis=0)
    np.save(save_path+'/example_importance_values.npy', example_shap_values)
    np.save(save_path+'/all_importance_values.npy', shap_values)
    shap_values = shap_values.reshape(-1, len(attris_name))
    X_trains = np.concatenate(pid_X_trains, axis=0).reshape(-1, len(attris_name))

    x_test_2d = pd.DataFrame(data=X_trains, columns = attris_name)  
    shap.summary_plot(shap_values, x_test_2d, show=False)
    plt.savefig(save_path+f'/{dataset}_shap.pdf')          
```
